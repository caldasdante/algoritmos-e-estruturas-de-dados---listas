# 1. Imagine que nós aplicamos o procedimento Partição() sobre a lista V e imagine que depois nós ordenamos os dois lados da partição usando o algoritmo da bolha.

a) Analise o tempo de execução desse algoritmo no melhor caso.

b) Analise o tempo de execução desse algoritmo no pior caso.


**RESPOSTAS:**

A) No mellhor caso, onde a lista está previamente ordenada e o pivô é o ultimo elemento da lista, o tempo de execução é O(n), já que a partição teria custo linear, e as ordenações de um lado seria O(n-1), lado com os elementos menores que o pivô previamente ordenados, e do outro O(1), lado com os elementos maiores que o pivô.

B) Já no pior caso, a partição seria desequilibrada e ambos lados estariam completamente desordenados, po essa lógica como a etapa de ordenação seria a mais custosa, no fim o tem po seria quadrático, ou seja, O(n^2).

# 2. Imagine que nós ordenamos os primeiros 2n/3 elementos da lista L usando o algoritmo dabolha. Daí, nós ordenamos os últimos 2n/3 elementos da lista usando o algoritmo da bolha, e finalmente, nós ordenamos os primeiros 2n/3 elementos da lista outra vez usando o algoritmo da bolha.

a) Ao final desse processo, a lista está completamente ordenada? Porque?

b) Analise o tempo de execução desse procedimento.

**RESPOSTAS:**

A)Sim, pois as duas regiões que são ordenadas (os primeiros `2n/3` elementos e os últimos `2n/3` elementos) se sobrepõem na região central da lista (entre `n/3` e `2n/3 - 1`). Juntas, elas cobrem toda a extensão da lista `[0, n-1]`. O algoritmo da bolha, ao ser aplicado repetidamente em segmentos que, em união, cobrem o array todo e se sobrepõem, permite que qualquer elemento se mova para sua posição correta, independentemente de sua posição inicial. A terceira passagem garante que os elementos que foram movidos para fora da primeira região na segunda passagem, agora podem ser realocados corretamente, consolidando a ordenação.

B) Tempo de execução O(n^2). Cada aplicação de Bubble Sort em 2n/3 elementos tem complexidade O((2n/3)^2) = O(n^2). Com três dessas operações, a complexidade total permanece O(n^2).

```python
import random
from random import randint

def creat_vetor(t, lim_inicio,lim_fim):
  vetor = [0] * t

  for i in range(t):
    vetor[i] = random.randint(lim_inicio,lim_fim)

  return vetor
```

```python
def cria_matriz(linhas, colunas):
  total_elements = linhas * colunas
  valor_unico = random.sample(range(1,8), total_elements)

  matriz = []
  for i in range(linhas):
        inicio = i * colunas
        fim =  inicio + colunas
        linha = valor_unico[inicio:fim]
        matriz.append(linha)
  return matriz
```

```python
c = cria_matriz(3,3)
for linha in c:
  print(linha)
```

```python
def check(matriz):
  buxca = set()
  for linhas in matriz:
    for element in linhas:
      if element in buxca:
        return True
      buxca.add(element)

if check(c):
  print("A matriz M contém elementos duplicados.")
else:
  print("A matriz M não contém elementos duplicados.")
```
