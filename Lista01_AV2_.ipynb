# 1. O maior no fim.

---

**a) Apresente um algoritmo que resolve esse problema.**

```python
def mov_fim(lista_ref):

    if not lista_ref or not lista_ref.proximo:
        return lista_ref

    maior_no = lista_ref
    corrente = lista_ref

    while corrente:
        if corrente.dado > maior_no.dado:
            maior_no = corrente
        corrente = corrente.proximo

    ultimo_no = lista_ref
    while ultimo_no.proximo:
        ultimo_no = ultimo_no.proximo

    if maior_no != ultimo_no:
        temporario = maior_no.dado
        maior_no.dado = ultimo_no.dado
        ultimo_no.dado = temporario

    return lista_ref


lista_ref = criar_lista_encad(10,1,100)
exibir_lista(lista_ref)

mov_fim(lista_ref)
exibir_lista(lista_ref)
```

**b) Analise o tempo de execução do seu algoritmo.**

O tempo de execução desse algoritmo é linear, pois são feitos 3 processos independentes, o primeiro de buscar o maior é feito em O(N), o segundo é a busca pelo último elemento, que também é linear, e o último processo é são as trocas sucessivas até o maior elemento ficar no final, com tempo constante. Desse modo o tempo de execução do algoritmo é linear.

```python
def inverter_lista(inicio_lista):
    anterior_no = None
    corrente = inicio_lista
    while corrente:
        proximo = corrente.proximo
        corrente.proximo = anterior_no
        anterior_no = corrente
        corrente = proximo
    return anterior_no

lista_invertida = criar_lista_encad(10,1,100)
print("A lista original é:")
exibir_lista(lista_invertida)

lista_final = inverter_lista(lista_invertida)
print("\nLista invertida é:")
exibir_lista(lista_final)
```

OBS: Como não há uma ideia muito clara de como deveria fucionar a duplicação dos ímpares optei por apenas duplicar os elementos que são ímpares e colocar os mesmos como próximos dos elementos originais.

```python
def duplicar_impares(lista_ref):
    corrente = lista_ref

    while corrente is not None:
        if corrente.dado % 2 != 0:
            novo = No(corrente.dado)
            novo.proximo = corrente.proximo
            corrente.proximo = novo
            corrente = novo.proximo
        else:
            corrente = corrente.proximo

    return lista_ref

atual_impar = criar_lista_encad(10,1,100)
print("A lista orginal é:")
exibir_lista(atual_impar)

duplicar_impares(atual_impar)
print("Lista atual é:")
exibir_lista(atual_impar)
```

```python
def intercalar(lista_um_ref, lista_dois_ref):
    resultado_no = No(0)
    corrente = resultado_no

    while lista_um_ref is not None and lista_dois_ref is not None:
        if lista_um_ref.dado <= lista_dois_ref.dado:
            corrente.proximo = lista_um_ref
            lista_um_ref = lista_um_ref.proximo
        else:
            corrente.proximo = lista_dois_ref
            lista_dois_ref = lista_dois_ref.proximo
        corrente = corrente.proximo

    if lista_um_ref is not None:
        corrente.proximo = lista_um_ref
    elif lista_dois_ref is not None:
        corrente.proximo = lista_dois_ref

    return resultado_no.proximo

lista_um = criar_lista_encad(5,1,100)
lista_dois = criar_lista_encad(5,1,100)

print("As listas são:")
exibir_lista(lista_um)
exibir_lista(lista_dois)

lista_tres = intercalar(lista_um,lista_dois)
print("\nLista intercalada:")
exibir_lista(lista_tres)
```

**b) Analise o tempo de execução do seu algoritmo.**

**a) Apresente um algoritmo que resolve esse problema.**

```python
def particionar(lista_ref, k):
    sentinela = No(0)
    fim_menores = sentinela
    inicio_maiores = fim_maiores = None

    corrente = lista_ref
    while corrente is not None:
        proximo = corrente.proximo
        corrente.proximo = None

        if corrente.dado <= k:
            fim_menores.proximo = corrente
            fim_menores = corrente
        else:
            if inicio_maiores is None:
                inicio_maiores = fim_maiores = corrente
            else:
                fim_maiores.proximo = corrente
                fim_maiores = corrente

        corrente = proximo

    fim_menores.proximo = inicio_maiores
    return sentinela.proximo

limite_maior = criar_lista_encad(10,1,100)
print("A lista original é:")
exibir_lista(limite_maior)

while True:
  k = int(input("\nDigite um número entre 1 e 100. "))
  if k<1 or k>100:
    print("Valor fora do intervalo.")
  else:
    break

limite_menor = particionar(limite_maior, k)
print(f"\nA lista particionada em {k} é:")
exibir_lista(limite_menor)
```

**b) Analise o tempo de execução do seu algoritmo.**

​O algoritmo realiza uma única passagem de início ao fim da lista (n iterações). Em cada nó, é feita uma verificação condicional simples e atualizações de ponteiros O(1) para direcioná-lo para a sublista correspondente. A junção final das duas listas também ocorre em tempo constante O(1). Por fim, o tempo de execução do algoritmo é O(n), onde n é o número total de nós da lista.

# 8. O elemento que aparece mais vezes.
---

image.png

**a) Apresente um algoritmo que resolve esse problema.**

```python
def mais_frequente(lista_ref):
    if lista_ref is None:
        return None

    valor_item = None
    maior_contagem = 0
    primeiro_no = lista_ref

    while primeiro_no is not None:
        contador = 0
        segundo_no = lista_ref

        while segundo_no is not None:
            if segundo_no.dado == primeiro_no.dado:
                contador += 1
            segundo_no = segundo_no.proximo

        if contador > maior_contagem:
            maior_contagem = contador
            valor_item = primeiro_no.dado

        primeiro_no = primeiro_no.proximo

    return valor_item, maior_contagem

lista_base = criar_lista_encad(10, 1, 100)
exibir_lista(lista_base)

elemento, contador = mais_frequente(lista_base)
print(f"O elemento mais frequente é o {elemento}, apareceu {contador} vezes")
```

**b) Analise o tempo de execução do seu algoritmo.**

Tempo de execução é quadrático O(N²), pois para cada elemento a lista é percorrida N vezes, sendo N o número de elementos, buscando elementos repetidos dentro dos dois laços While.

# 9. Elemento repetido.

**a) Apresente um algoritmo que resolve esse problema.**

```python
def tem_repetido(lista_ref):
    primeiro_no = lista_ref
    while primeiro_no is not None:
        segundo_no = primeiro_no.proximo
        while segundo_no is not None:
            if primeiro_no.dado == segundo_no.dado:
                return f"Sim, o {primeiro_no.dado}"
            segundo_no = segundo_no.proximo
        primeiro_no = primeiro_no.proximo
    return "Não possui"

lista_exemplo = criar_lista_encad(10, 1, 100)
exibir_lista(lista_exemplo)

resultado = tem_repetido(lista_exemplo)
print(resultado)
```

**b) Analise o tempo de execução do seu algoritmo.**

O(N²). O algoritmo compara cada elemento da lista com todos os seus sucessores usando dois loops encadeados. No pior caso, realiza cerca de N²/2 comparações.

**a) Apresente um algoritmo que resolve esse problema.**

```python
def intersecao(lista_um_ref, lista_dois_ref):
    sentinela = No(0)
    ultimo_resultado = sentinela

    primeiro_no = lista_um_ref
    while primeiro_no is not None:
        segundo_no = lista_dois_ref
        encontrado = False
        while segundo_no is not None:
            if primeiro_no.dado == segundo_no.dado:
                encontrado = True
                break
            segundo_no = segundo_no.proximo

        if encontrado:
            novo = No(primeiro_no.dado)
            ultimo_resultado.proximo = novo
            ultimo_resultado = novo

        primeiro_no = primeiro_no.proximo

    return sentinela.proximo

print("Listas originais:")
lista_a = criar_lista_encad(10, 1, 100)
lista_b = criar_lista_encad(10, 1, 100)

exibir_lista(lista_a)
exibir_lista(lista_b)

print("\nLista de intersecção:")
pão = intersecao(lista_a, lista_b)
exibir_lista(pão)
```

**b) Analise o tempo de execução do seu algoritmo.**

​A complexidade do algoritmo é O(N×M). Para cada um dos N elementos da primeira lista, o algoritmo percorre até os M elementos da segunda lista para verificar se há interseção. O tempo total é proporcional ao produto do tamanho das duas listas.
