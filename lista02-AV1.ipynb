```python
import random
from random import randint

def creat_vetor(t, lim_inicio, lim_final):
  vetor = [0] * t

  for i in range(t):
    vetor[i] = random.randint(lim_inicio, lim_final)

  return vetor
```

```python
def ordem(v):
  for i in range(len(v)):
    for j in range(i+1, len(v)):
      if v[i] > v[j]:
        aux = v[i]
        v[i] = v[j]
        v[j] = aux
  return v
```

# 1. O terceiro maior elemento

```python
# CRIANDO VETOR
mot = creat_vetor(10,1,100)
print("Vetor inicial:\n", mot)

# ODENANDO VETOR
n = len(mot)
i = 0

while i < n - 1:
  j = 0
  while j < n - 1 - i:
    if mot[j] > mot[j+1]:
      aux = mot[j]
      mot[j] = mot[j+1]
      mot[j+1] = aux
    j += 1
  i += 1

print("Vetor ordenado:\n", mot)
print("O maior elemento da lista é:", mot[len(mot)-1])
```

# 2. k-ésimo maior elemento

```python
ker = creat_vetor(10,1,100)
print("O vetor é:\n", ker)
ordem(ker)
print("O vetor ordenado é:\n", ker)

k =  int(input("\nDigite um valor entre 1 e 10:"))

print(f"O {k}° maior elemento é:", ker[len(ker) - k])
```

# 3. O mais próximo da média

```python
mid =  creat_vetor(10,1,100)
print("O vetor é:\n", mid)

i = 0
soma = 0
for i in range(len(mid)):
  soma = soma + mid[i]
media = soma/len(mid)
print(f"A  média do veto é {media}.")

proxMed = abs(mid[0] - media)
prox = mid[0]

i = 0
for i in range(len(mid)):
  atMedia = abs(mid[i] - media)
  if atMedia < proxMed:
    proxMed =  atMedia
    prox = mid[i]

print(f"O elemento mais próximo da média é: {prox}")
```

# 4. Elemento isolado

```python
iso = creat_vetor(10,1,10)
print("O vetor é:\n", iso)
ordem(iso)
print("O vetor ordenado é:\n", iso)

achei = 0
for i in range(1,len(iso)-1):
  if iso[i-1]!=(iso[i]-1) and iso[i+1]!=(iso[i]+1):
    print(f"Achei, o elemento isolado é o {iso[i]}, na posição {i+1}.")
    achei = 1
    break
if achei == 0:
  print("Não há um elemento isolado")
```

# 5. k repetições

```python
rep = creat_vetor(10,1,10)
print("O vetor é:", rep)
k =  int(input("Digite um número entre 1 e 10: "))

achei = 0
i = 0
for i in range(len(rep)):
  if achei!=0:
    break
  cont = 0
  for j in range(i+1, len(rep)):
    if rep[i] == rep[j]:
      cont+=1
  if cont >= k:
    print(f"O elemento que aparece ao menos {k} vezes é o {rep[i]}.")
    achei = 1
    break

if achei==0:
  print(f"Não há elemento que apareça ao menos {k} vezes.")
```

# 6. Permutações

```python
tuc = creat_vetor(10,1,100)
print("O vetor 1 é:\n", tuc)
cut = creat_vetor(10,1,100)
print("O vetor 2 é:\n", cut)

ordem(tuc)
ordem(cut)

i = 0
y = 0
certo = 0
if len(tuc) != len(cut):
  certo = 1
  print("Uma não é permutação da outra (tamanhos diferentes)")
else:
  for i in range (len(tuc)):
    if certo == 0:
      if tuc[i] == cut[y]:
        y+=1
      else:
        certo = 1
    else:
      print("Uma não é permutação da outra")
      break

  if certo == 0:
    print("Uma é permutação da outra")
```

# 7. Repetidos próximos

```python
pet = creat_vetor(10,1,10)
print("O vetor é:\n", pet)
k = int(input("Digite até que distância quer saber se há dois elementos repetidos: "))

i = 0
ache = 0
for i in range(len(pet)):
  if ache != 0:
    break
  else:
    for j in range(i+1, (len(pet))):
      if pet[i] == pet[j] and (j-i <= k):
        print(f"Achei. O {pet[i]} está nas posições {i} e {j}")
        ache = 1
        break
if ache ==0:
  print(f"Não há elementos repetidos com no máximo {k} posição(ões) de distância.")


```
